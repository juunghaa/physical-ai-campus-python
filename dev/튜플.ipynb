### 튜플 자료형

튜플 : 불변(immutable) 특징, 리스트와의 차이, 상수, 집합 등에 사용 

```python
# 튜플 예시

t1 = ()
t2 = (1, )
t3 = (1,2,3)
t4 = 1,2,3
t5 = ("a", "b", ("ab", "cd"))
```

```python
t4
```

```python
t5[2][0]
```

```python
# 튜플의 특성 (요소값 삭제, 변경이 안됨)

t1 = (1,2,'a','b')
del t1[0]
```

```python
t1 = (1,2,'a','b')
t1[0] = 'c'
```

```python
# 인덱싱 하기 

t1 = (1,2,'a','b')
t1[0]
```

```python
# 슬라이싱 하기

t1 = [1,2,'a','b']
t1[1:]
```

```python
# 튜플 더하기

t1 = (1,2,'a','b')
t2 = (3,4)

t3 = t1+t2
t3
```

```python
# 튜플 곱하기

t2 = 3,4
t3 = t2*3

t3
```

```python
# 튜플 길이 구하기

t1 = [1,2,'a','b']
len(t1)
```

```python
# 리스트는 데이터 변경 시 많이 사용
# 튜플은 데이터 변경 못하게 할 때 사용함(상수값 같은 거)

# 튜플은 요소 값을 변경할 수 없기 때문에 sort, insert, remove, pop 메서드가 없음 
```
