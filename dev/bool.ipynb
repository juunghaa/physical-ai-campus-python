```python
# 자료형 참과 거짓 (문자열, 리스트, 튜플, 딕셔너리 등의 값이 비어 있으면 거짓이고 비어 있지 않으면 참이 된다.)

a = [1,2,3,4]
while a: # while 조건문(a 리스트가 요소가 있으니까 true가 됨)
    print(a.pop()) # 뒤에서 하나씩 꺼내는데, 빼면 사라짐 

# while문은 조건이 False가 되는 순간 끝남 
```

```python
if [] : # 리스트가 비어있어 False가 됨 
    print("참")
else :
    print("거짓")
```

```python
if [1,2] : 
    print("참")
else :
    print("거짓")
```
