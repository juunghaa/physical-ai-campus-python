```python
# 실습 : 간단한 계산기 만들기

print("=== 간단한 계산기 ===")

# 사용자로부터 두 숫자 입력 받기

a = float(input("1th 숫자를 입력하세요 : "))
b = float(input("2th 숫자를 입력하세요 : "))

# 계산식

print(f"두 수의 합 : {a+b}\n두 수의 차 : {a-b}\n두 수의 곱 : {a*b}")

if b != 0 :
    print(f"두 수의 나눗셈 : {a/b}\n")
else : 
    print(f"0으로 나눌 수 없습니다.")
```
