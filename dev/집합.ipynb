### 집합 자료형

집합 : 중복 제거, 교집합/합집합/차집합 연산, 순서 없음

```python
# 집합 자료형 예제

s1 = set([1, 2, 3, 3, 3])
s1
```

```python
s2 = set("Hello") # 중복된 단어가 하나만 들어감 
s2
```

```python
s3 = {1,2,3,3,3}
s3
```

```python
s4 = {'a','b','c','b','c'}
s4
```

```python
# 집합 자료형을 리스트형으로 변환

list(s4)
```

```python
tuple(s4)
```

```python
s4[0] # 집합 자료형은 순서가 없기 때문에 인덱싱이 되지 않음 
```

```python
# 교집합, 합집합, 차집합 

s1 = set([1,2,3,4,5,6])
s2 = set([4,5,6,7,8,9])
```

```python
# 교집합 

s1 & s2
```

```python
s1.intersection(s2)
```

```python
# 합집합

s1 | s2
```

```python
s1.union(s2)
```

```python
# 차집합 

s1 - s2
```

```python
s2 - s1
```

```python
s1.difference(s2)
```

```python
s2.difference(s1)
```

```python
# 값 1개 추가하기 - add

s1 = {1,2,3}
s1.add(4)
s1
```

```python
# 값을 여러개 추가하기 - update

s1 = {1,2,3}
s1.update([4,5,6])
s1
```

```python
# 특정 값 제거하기 - remove

s1 = {1,2,3}
s1.remove(2)
s1
```

```python
# 특정 값 제거하기 - discard

s1 = {1,2,3}
s1.discard(2)
s1
```

```python
s1.discard(4) # 4가 없음에도 오류가 나지 않음 
```

```python
s1.remove(4) # 오류가 발생함 
```

```python
# discard를 왜 사용하는가?
##### -> 위와 같이.. discard는 없는 값이어도 오류가 발생하지 않음 
######## -> 원하는 결과가 나오도록 discard와 remove를 적절히 사용하는 것 
```

```python
# 모든 값을 제거하기 - clear

s1 = {1,2,3}
s1.clear()
s1
```
