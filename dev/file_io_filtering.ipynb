```python
# 여러 조건을 동시에 사용하여 필터링을 진행
# pandas에서 and 조건은 &, or 조건은 | 사용
# 조건을 여러개 사용시 괄호를 넣어야 함 

line_a_ng = df_json[(df_json["line"]=="Line_A") & (df_json["status"]=="NG")]  # Line A와 NG 조건을 필터링 

line_a_ng.head() 
```

```python
# between()으로 특정 범위에 들어오는 데이터를 필터링 
# 압력이 2.7 이상 3.0 이하인 데이터를 필터링 하기 

normal_pressure_range = df_json[df_json["pressure"].between(2.7, 3.0)]

normal_pressure_range.head()
```

```python
normal_pressure_range["pressure"].describe()
```

```python
# 데이터 정렬하기 

# quality_score (품질 점수)가 낮은 데이터부터 정렬하기 

low_quality_score = df_json.sort_values(by="quality_score", ascending=True) # ascending은 올림차순 정렬 

low_quality_score.head(10)
```

```python
# 여러 기준으로 정렬하기 
# line 기준으로 정렬하고, 같은 line 안에서는 timestamp 기준으로 정렬 

line_time_sorted = df_json.sort_values(by=["line", "machine_id"], ascending=[True, True])
line_time_sorted.head(10)
```

```python
# groupby 로 요약하기 

# line별 데이터 개수와 평균 센서값을 요약하기 

line_summary = df_json.groupby("line").agg(
    # 수집하는 정보를 입력
    data_count = ("timestamp", "count"),  # 대상 컬럼, 평균/개수/최대/최소
    avg_temperature = ("temperature", "mean"),
    avg_pressure = ("pressure", "mean"),
    avg_speed = ("speed", "mean"),
    avg_quality_score = ("quality_score", "mean"),
).reset_index()

line_summary
```

```python
# 설비별 OK/NG 개수를 계산하기 

machine_status_summary = df_json.groupby(["machine_id", "status"]).agg(
    count = ("timestamp", "count")
).reset_index()

machine_status_summary
```

```python
# pivot_table을 사용하면 설비별 OK/NG 개수를 표 형태로 보기 좋게 만들 수 있음 

machine_status_pivot = pd.pivot_table(
    df_json,    # 대상 데이터프레임 지정 
    index = "machine_id", # 인덱스 컬럼 
    columns = "product", # 컬럼 (OK, NG 가 나뉘어 컬럼으로 들어가게 됨)
    values = "pressure", # 실제 피봇에서 사용할 데이터 
    aggfunc = "mean", # 수집할 함수 (개수, 평균/최대/최소/표준편차)
    fill_value = 0, # 결측치는 0으로 채우겠다는 것 (기본 값 설정)
).reset_index()

machine_status_pivot
```

```python
# 제품별 평균 품질 점수와 NG 비율 계산하기

product_report = df_json.groupby("product").agg(
    total_count = ("status", "count"), # 전체 개수 집계
    ng_count = ("status", lambda x:(x=="NG").sum()), # ng에 해당되는 개수 집계
    avg_quality_score = ("quality_score", "mean"),
    avg_temperature = ("temperature", "mean"),
    avg_vibration = ("vibration", "mean")
).reset_index()

product_report
```

```python
# NG 비율 계산 컬럼을 추가하기

product_report["ng_ratio"] = product_report["ng_count"] / product_report["total_count"] * 100
product_report = product_report.round(2)

product_report
```

```python
# 리포트 파일을 저장하기

report_path = r"C:\Users\user\Desktop\python_campus\dev\data\result\quality_report.csv"

product_report.to_csv(report_path, index=False, encoding="utf-8-sig")
```

line별 NG 개수와 NG 비율을 계산하세요.

```python
line_ng_report = df_json.groupby("line").agg(
    totalCount = ("status", "count"),
    ngCount = ("status", lambda x: (x=="NG").sum()),
).reset_index()

line_ng_report["ng_ratio"] = line_ng_report["ngCount"] / line_ng_report["totalCount"] * 100
line_ng_report = line_ng_report.round(2)

line_ng_report
```
