### 함수

함수 : 정의와 호출, 매개변수 종류 (위치, 기본값, 키워드, 가변), 반환값

```python
# 함수 예시

def add(a, b):
    return a+b
```

```python
# 함수의 사용

a = 3
b = 4
add(a,b)
```

```python
# 매개변수와 인수
# parameter = 매개변수 : 함수의 입력으로 전달되는 값을 받는 변수 
# argument = 인수 : 함수를 호출할 때 전달되는 입력 값
```

```python
def add (a,b) : # 매개변수
    return a+b

print(add(3,4)) # 인수 
```

```python
# 일반적인 함수

def add (a,b) :
    result = a+b
    return result

a = add(3,4)
a
```

```python
# 입력값이 없는 함수

def say():
    return "hi"

a = say()
a
```

```python
# 반환값이 없는 함수 (return문이 없음)

def add(a,b) :
    print(f"{a}, {b}의 합은 {a+b}입니다.")

add(3,4)
```

```python
# 매개변수를 지정해서 호출하기

def sub(a,b) :
    return a-b
```

```python
result = sub(a=7, b=3)
print(result)
```

```python
result = sub(b=5, a=3) # 매개변수 지정시 순서는 상관 없음
print(result)
```

```python
# 여러 개의 입력값을 받는 함수 만들기

def add_many(*args) : # *args는 여러개의 입력을 받는 매개변수임 
    result = 0
    for i in args: 
        result += i
    return result
```

```python
result = add_many(1,2,3)
result
```

```python
def add_mur(choice, *args):
    if choice == "add" :
        result = 0
        for i in args:
            result += i
    elif choice == "mul":
        result = 1
        for i in args:
            result *= i
    return result
```

```python
result = add_mur("add", 1,2,3,4,5)

result
```

```python
result = add_mur("mul", 1,2,3,4,5)

result
```

```python
# 키워드 매개변수, **kwargs (키워드 매개변수는 함수 호출시 키워드=값 형태로 전달하는 매개변수를 받을 때 사용)

def print_kwargs(**kwargs) :
    print(kwargs)
```

```python
print_kwargs(a=1) # 딕셔너리 형태로 출력 (key : value)
```

```python
print_kwargs(name="foo", age=3)
```

```python
print_kwargs(name="홍길동", age=25, city="서울", job="개발자")
```

```python
# 키워드 매개변수 예시

def create_profile(**info):
    print("=== 프로필 정보 ===")
    for key, value in info.items(): # info의 key value 쌍을 가져오는 함수 = items()
        print(f"{key} : {value}")
```

```python
create_profile(name="gildong", age = 30, job = "developer", hobby = "reading")
```

```python
# 일반 매개변수, 가변 매개변수, 키워드 매개변수 함께 사용 예시

def mixed_func(name, *args, **kwargs):
    print(f"name : {name}")
    print(f"추가 인수들 : {args}")
    print(f"키워드 인수들 : {kwargs}")
```

```python
mixed_func('gildong', 1,2,3, age=25, city='seoul')
```

```python
# 함수의 반환 값은 언제나 하나이다.

def add_and_mul(a,b):
    return a+b, a*b # 튜플 값으로 나옴 
```

```python
# 함수 안에서 선언한 변수의 효력 범위

a = 1
def vartest(a) :
    a = a+1 # 함수 안에서의 변수는 함수 안에서만 유효함 

vartest(a) # 함수 안의 a는 2이지만, 함수 밖의 a는 1임 
print(a) # 함수 밖에 있는 a를 의미함 
```

```python
# 함수 안에서 함수 밖의 변수를 변경하는 방법

a = 1
def vartest(a):
    a+=1
    return a

a = vartest(a)
print(a)
```

```python
# global 명령어 사용하기 (global 변수는 함수 안에서 함수 밖의 변수를 직접 사용하겠다는 의미)

b = 1
def vartest():
    global b # 함수 밖의 매개변수를 그대로 사용하겠다는 의미
    b += 1

vartest()
print(b)
```

```python
# 리스트나 딕셔너리는 함수에서 변경이 가능함

def change_list(my_list) :
    my_list.append(4)

a = [1,2,3]
change_list(a)
print(a)
```

```python
# lambda : def와 동일한 역할을 함. 보통 함수를 한 줄로 간결하게 만들 때 사용함

add = lambda a,b : a+b # lambda 매개변수, 매개변수 ... : 매개변수를 이용한 표현식

result = add(3,4)
result
```

```python
# 함수의 독스트링 (Docstring) : 함수에 대한 설명을 문서화하는 방법
# 함수의 첫번째 줄에 삼중 따옴표로 작성함 

def add(a,b) :
    """
    두 숫자를 더하는 함수 

    Parameter: 
    a (int, float) : 1th 숫자
    b (int, float) : 2th 숫자

    Return:
    int, float : 두 숫자의 합
    """
    return a+b
```
