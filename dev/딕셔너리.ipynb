### 딕셔너리 자료형

딕셔너리 : 키(Key) - 값(Value) 구조, 접근/추가/삭제, 키의 해시 가능성 유의점 (키가 중복되면 안됨)

```python
# 딕셔너리 예제 {key1 : value1, key2 : value2, ...}

dic = {'name':'pey', 'phone':'010-0000-0000', 'birth':'1118'}
dic
```

```python
type(dic)
```

```python
a = {1:'hi'}
a
```

```python
a = {'a':[1,2,3]}
a
```

```python
# 딕셔너리 쌍 추가하기

a = {1:'a'}
a[2] = 'b' # 키 값이 추가 됨 
a
```

```python
a['name']='pey'
a
```

```python
a[3]=[1,2,3]
a
```

```python
# 딕셔너리 요소 삭제하기 

del a[1] # 키와 밸류가 같이 삭제 됨 
a
```

```python
# 딕셔너리 사용하기 

{"김연아":"피겨스케이팅", "류현진":"야구", "손흥민":"축구", "귀도":"파이썬"}
```

```python
# 딕셔너리에서 key를 사용하여 value 얻기

grade = {"pey":10, "julliet":90}
grade["pey"]
```

```python
dic = {'name':'pey', 'phone':'010-0000-0000', 'birth':'1118'}

dic['name']
```

```python
# 딕셔너리 만들 때 주의사항 (Key는 고유한 값이어야 함, 중복되면 데이터가 무시됨)

a = {1:'a', 1:'b'}
a
```

```python
# 키 값으로 사용하면 안되는 것 

a = {[1,2]:'hi'} # 키 값으로 리스트는 사용할 수 없음, 유일한 값이어야 하므로!
```

```python
a = {'name':'pey', 'phone':'010-0000-0000', 'birth':'1118'}
a.keys()
```

```python
a.values()
```

```python
# 키 값들을 하나씩 뽑아서 k에 담음 
for k in a.keys():
    print(k)
```

```python
list(a.keys()) # 리스트로 변환해줌 
```

```python
# value 리스트 만들기

list(a.values())
```

```python
# key와 value 쌍 얻기 

a.items()
```

```python
list(a.items())
```

```python
# 키, 밸류 쌍 모두 지우기

a.clear()
a
```

```python
# 키로 밸류 얻기 - get

a = {'name':'pey', 'phone':'010-0000-0000', 'birth':'1118'}
a.get("name")
```

```python
a["name"]
```

```python
a.get('nokey')
```

```python
a.get('nokey', '정보없음')
```

```python
# 해당 키가 딕셔너리 안에 있는지 조회하기 - in

a = {'name':'pey', 'phone':'010-0000-0000', 'birth':'1118'}

'name' in a # a안에 name이 있는가?
```

```python
'email' in a
```

```python
# 키로 밸류 꺼내기 - pop

a = {'name':'pey', 'phone':'010-0000-0000', 'birth':'1118'}
phone = a.pop('phone') #키값에 해당하는 값을 꺼냄 (꺼내면 꺼낸 대상이 사라짐)
phone
```
