### if문

조건문 : if/elif/else 구조, 중첩과 가독성 팁, 짧은 조건시 예시

```python
# if문 예시 (주어진 조건을 판단 후 그 상황에 맞게 처리)

money = True
if money : # 조건이 참이면 
    print("taxi")
else : 
    print("walk")
```

```python
# 다양한 조건을 판단하는 elif

pocket = ["paper", "cellphone"]
card = True

if "money" in pocket :
    print("taxi")
else : 
    if card : 
        print("bus")
    else : 
        print("walk")
```

```python
pocket = ["paper", "cellphone"]
card = True

if "money" in pocket :
    print("taxi")
elif card :
    print("bus")
else :
    print("walk")
```

```python
# match-case 문

grade = "B"
if grade == "A" :
    print("excellent")
elif grade == "B" :
    print("good")
elif grade == "C" : 
    print("not bad")
else : 
    print("bad")
```

```python
grade = "D"
match grade : 
    case "A" :
        print("excellent")
    case "B" :
        print("good")
    case "C" :
        print("not bad")
    case _ :
        print("bad")
```

```python
# 여러 패턴을 하나의 case에서 처리하기 

grade = "B"
match grade :
    case "A" | "B" | "C" :
        print('pass')
    case _ :
        print('fail')
```

점수가 90점 이상이면 A, 80점 이상이면 B, 그렇지 않으면 C를 출력하는 코드를 작성하세요.

```python
score = 85

if score >= 90 : print("A")
elif score >= 80 : print("B")
else : print("C")
```

변수 year가 2024일 때 윤년인지 아닌지 판별하는 코드를 작성하세요. (윤년: 4로 나누어 떨어지고 100으로 나누어 떨어지지 않거나 400으로 나누어 떨어짐)

```python
year = 2024

if year%4==0 and year%100!=0 or year%400==0 : print("윤년")
else : print("평년")
```
